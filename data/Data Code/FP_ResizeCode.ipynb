```python
import cv2
import os
import shutil
from zipfile import ZipFile

# Function to resize the image to 512x927 while preserving the filename as PNG
def resize_image(input_image_path, output_directory, target_width=512, target_height=927):
    # Read the image
    image = cv2.imread(input_image_path)

    if image is None:
        print(f"Error: Failed to load the image: {input_image_path}")
        return

    # Resize the image to 512x927
    resized_image = cv2.resize(image, (target_width, target_height))

    # Get the filename and change extension to .png
    filename = os.path.basename(input_image_path)
    filename_without_extension = os.path.splitext(filename)[0]
    filename_png = filename_without_extension + ".png"

    # Save the resized image
    output_path = os.path.join(output_directory, filename_png)
    cv2.imwrite(output_path, resized_image)
    print(f"Resized image saved to {output_path}")

    return output_path

# Input and output directories
input_directory = "dataset"
output_directory = "resized_images"

# Create output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# List to hold resized image paths
resized_images_paths = []

# Loop through all files in the dataset folder
for filename in os.listdir(input_directory):
    file_path = os.path.join(input_directory, filename)
    if os.path.isfile(file_path):
        resized_path = resize_image(file_path, output_directory)
        if resized_path:
            resized_images_paths.append(resized_path)

print(f"All images resized and saved in '{output_directory}/'.")
```
